```python
import numpy as np
from glaciation_time_estimator.auxiliary_func.chunking_data import ChunkLoader
```

```python
import pandas as pd
```

```python
chunker_weighted = ChunkLoader([2008],config_fp="/wolke_scratch/dnikolo/Glaciation_time_estimator/configs/debug/debug.yaml")
chunker_non_weighted = ChunkLoader([2008],config_fp="/wolke_scratch/dnikolo/Glaciation_time_estimator/configs/debug/debug_non_weighted.yaml")
```

```python
clouds_weighted = chunker_weighted.cloud_chunk
clouds_non_weighted = chunker_non_weighted.cloud_chunk
```

```python
def occ_rate(cloud_chunk):
    return pd.pivot_table(cloud_chunk[cloud_chunk.max_water_frac > 0.4],values='is_glaciating' ,index=["Hemisphere"], columns=["min_temp"], aggfunc='mean')
```

```python
table_weighted = occ_rate(clouds_weighted)
table_non_weighted = occ_rate(clouds_non_weighted)
```
